### Setting up the Environment 

```python
import os
from dotenv import load_dotenv
import json

load_dotenv()

openai_api_key = os.getenv("OPENAI_API_KEY")
```

### Create the OpenAI Client

```python
from openai import OpenAI

client = OpenAI(
    api_key = openai_api_key
)
```

### Vectorize the dataset

```python
with open("final_rag_dataset.json", "r", encoding="utf-8") as f:
    documents = json.load(f)

for document in documents:

    response = client.embeddings.create(
        model = "text-embedding-3-small",
        input = document["chunk"]
    )

    document["vector"] = response.data[0].embedding

with open("vector-data.json", "w", encoding="utf-8") as f:
    json.dump(documents, f, indent=4, ensure_ascii=False)

print(f"Generated embeddings for {len(documents)} documents.")
```
